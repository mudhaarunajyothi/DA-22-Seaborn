# SEABORN

```python
# Def: Seaborn is a high level visualization library built on the top of the Matplotlib.

# Seaborn library popularily used for plotting Statistical charts.(eg: Box plot, KDE plot, pair plot)

# Seaborn provides inbulit datasets also.

# Matplotlib provide detailed control on the chart while Seaborn don't provide detailed control on chart.
```

```python
import seaborn as sns
```

```python
# datasets inside the seaborn
print(sns.get_dataset_names())
```

```python
# titanic dataset

df = sns.load_dataset("titanic")
df
```

```python
# flights dataset
df = sns.load_dataset("flights")
df
```

# Line chart

# All the charts of matplotlib can also be plotted by seaborn

```python
# trends over the time

# on the same chart features of seaborn on matplotlib can be apply

import matplotlib.pyplot as plt

timeline = df.groupby("year")["passengers"].sum().reset_index()

sns.lineplot(x = "year", y = "passengers", data = timeline , marker = "o")
plt.title("Year wise  passengers")
plt.grid()
```

```python
# Scatter plot

df = sns.load_dataset("tips")
df
```

```python
# to find the relation between two variables

plt.figure(figsize = (3,3))

sns.scatterplot(x = "total_bill", y = "tip", data = df )
```

```python
# to find the relation between two variables

plt.figure(figsize = (3,3))

sns.scatterplot(x = "total_bill", y = "tip", data = df, hue = "smoker")
```

# Pair plot

```python
# Pair plot : compare all possible  numerical pair of variable at a time

plt.figure(figsize = (4,3))
# compare multiple columns with each other together
sns.pairplot(df,hue = "smoker")                       #
```

# HeatMap

```python
# USE : heatmap visualize the EXACT correlation between numerical/categorical variables.

# Heatmap is used for plotting correlation matrix in colorful chart

import pandas as pd

# df.corr() function is used to generate correlation table
corr = df.corr(numeric_only = True)

print(corr)
```

```python
sns.heatmap(corr, annot = True)
```

```python
sns.heatmap(corr, annot = True, cmap = "coolwarm", fmt = ".3f") # we can change color  by changing cmap ="Greens", "Reds","YlGnBu","Purples"
```

```python
sns.heatmap(corr, annot = True, cmap = "Purples", fmt = ".3f")
```

# Heatmap for categorical variable

```python
# Finding relation between categorical variable

# numerical relation: heatmap + df.corr()

# categorical relation: heatmap + pivot table

# df.corr() function is used for finding relation between numerical variables
# Pivot table is used for finding relation between categorical variables.

# Heatmap is used just to show the relation tables in colorful chart
```

```python
pivot = df.pivot_table(index = "day", columns = "sex", aggfunc = "size", observed = False)

pivot
```

```python
pivot = df.pivot_table(index = "day", columns = "sex", aggfunc = "size", observed = False)

sns.heatmap(pivot, annot = True, cmap = "coolwarm")
```

```python
# Example-2:

#  size here is not the column name
# size: calculate total elements category wise
pivot = df.pivot_table(index = "day", columns = "smoker",values = "total_bill", aggfunc = "mean", observed = False)

plt.figure(figsize = (4,3))
sns.heatmap(pivot, annot = True)
```

```python
# USE: used to plot the linear regression line
plt.figure(figsize = (3,2))
sns.regplot(x = "total_bill", y = "tip", data = df)
```

# FacetGrid Plot

```python
# USE: to plot multiple charts based upon classes in categorical variable
```

```python
plt.figure(figsize = (2,1))
obj = sns.FacetGrid(df, col = "smoker")
obj.map_dataframe(sns.scatterplot, x = "total_bill", y = "tip")

plt.show()
```

```python
plt.figure(figsize = (2,1))
obj = sns.FacetGrid(df, col = "day")
obj.map_dataframe(sns.kdeplot, x = "total_bill", y = "tip")

plt.show()
```
